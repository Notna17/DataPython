## Індексно- та мітко-орієнтований вибір (`iloc`, `loc`)
[СКРІН 3] Після виконання комірок 10-16 зробіть скріншот прикладів `iloc`.
[СКРІН 4] Після виконання комірок 17-20 зробіть скріншот прикладів `loc`.

```python
# A) Вибір рядка на основі індексу
df.iloc[0]
```

```python
# Б) Вибір стовпця за індексом
df.iloc[:, 1] if df.shape[1] > 1 else df.iloc[:, 0]
```

```python
# В) Зріз елементів обраного стовпця
col_idx = 1 if df.shape[1] > 1 else 0
df.iloc[:6, col_idx]
```

```python
# Г) Зріз рядків
df.iloc[1:6]
```

```python
# Д) Зріз рядків з кінця за від'ємним індексом
df.iloc[-6:]
```

```python
# A) loc: вибір елемента за індексом і назвою стовпця
first_col = df.columns[0]
df.loc[0, first_col]
```

```python
# Б) loc: вибір підмножини стовпців за списком
subset_cols = list(df.columns[:3])
df.loc[:, subset_cols]
```

```python
# В) loc: перші 6 рядків з підмножини стовпців
subset_cols = list(df.columns[:3])
df.loc[:5, subset_cols]
```

```python
# Г) loc: вибір рядків з кроком у діапазоні
subset_cols = list(df.columns[:3])
stop_idx = min(len(df) - 1, 20)
df.loc[0:stop_idx:2, subset_cols]
```

```python
# Маніпулювання індексом: встановлюємо індекс за першим текстовим стовпцем
text_cols = df.select_dtypes(include=["object", "string"]).columns.tolist()
index_col = text_cols[0] if text_cols else df.columns[0]
df_by_name = df.set_index(index_col)
df_by_name.head()
```

```python
# Проста умова: значення числового стовпця більше за середнє
num_cols = df.select_dtypes(include=np.number).columns.tolist()
main_num_col = num_cols[0]
num_mean = df[main_num_col].mean()
print(f"Середнє для {main_num_col} = {num_mean:.4f}")
df.loc[df[main_num_col] > num_mean].head()
```

```python
# Складна умова з &, |
num_cols = df.select_dtypes(include=np.number).columns.tolist()
col_a = num_cols[0]
col_b = num_cols[1] if len(num_cols) > 1 else num_cols[0]
mean_a = df[col_a].mean()
median_b = df[col_b].median()

df.loc[(df[col_a] > mean_a) & (df[col_b] >= median_b)].head()
```

```python
# Умовні селектори: isin
cat_cols = df.select_dtypes(exclude=np.number).columns.tolist()
cat_col = cat_cols[0] if cat_cols else df.columns[0]
top_values = df[cat_col].dropna().astype(str).unique()[:2].tolist()
print("Колонка для isin:", cat_col)
print("Список значень:", top_values)
df.loc[df[cat_col].astype(str).isin(top_values)].head()
```

```python
# Умовні селектори: isnull / notnull
null_counts = df.isnull().sum()
target_null_col = null_counts.idxmax() if null_counts.max() > 0 else df.columns[0]
print("Колонка для перевірки null:", target_null_col)
print("isnull:")
display(df.loc[df[target_null_col].isnull()].head())
print("notnull:")
df.loc[df[target_null_col].notnull()].head()
```

```python
# Редагування даних: присвоєння сталого значення в новий стовпець
df["source_flag"] = "lab2"
df[[df.columns[0], "source_flag"]].head()
```

```python
# Редагування даних за правилом (умовне оновлення)
num_cols = df.select_dtypes(include=np.number).columns.tolist()
score_col = num_cols[0]
mean_score = df[score_col].mean()
df["score_level"] = np.where(df[score_col] >= mean_score, "high", "low")
df[[score_col, "score_level"]].head()
```

```python
# Редагування діапазону значень
df.loc[:5, "source_flag"] = "lab2_updated"
df[[df.columns[0], "source_flag"]].head(8)
```
