# Завантаження даних

```python
import pandas as pd
import numpy as np
# Завантажено з https://www.kaggle.com/datasets/ahmedmohamed2003/cafe-sales-dirty-data-for-cleaning-training
cafe_data = pd.read_csv("dirty_cafe_sales.csv")

# Зерно для генератора випадкових чисел
np.random.seed(17) 
```

### Визначення кількості відсутніх даних

```python
missing_values_count = (cafe_data.isnull() | cafe_data.isin(['UNKNOWN', 'ERROR'])).sum()
missing_values_count[0:8]
```

```python
# Визначаємо загальну кількість відсутніх елементів
total_cells = cafe_data.shape[0] * cafe_data.shape[1]
total_missing = missing_values_count.sum()
print(total_missing)

# Визначаємо відсоток відсутніх елементів
percent_missing = (total_missing/total_cells) * 100
print(percent_missing)
```

### Видалення відсутніх елементів даних

```python
# видаляємо всі рядки, що містять відсутні значення
cafe_data = cafe_data.replace(['UNKNOWN', 'ERROR'], None)
cafe_data.dropna()
```
