```python
# mean() для числових стовпців
print("Середній Fare:", df["Fare"].mean())
print("Середній Age:", df["Age"].mean())
```

```python
# unique() для рядкового стовпця
df["Embarked"].dropna().unique()
```

```python
# value_counts() для категоріальних значень
df["Embarked"].value_counts(dropna=False)
```

```python
fare_mean = df["Fare"].mean()
fare_centered_map = df["Fare"].map(lambda x: x - fare_mean)
fare_centered_map.head()
```

```python
def transform_row(row):
    # Центруємо Fare та підвищуємо ціну на 20%
    row["Fare_centered"] = row["Fare"] - fare_mean
    row["Fare_plus_20pct"] = row["Fare"] * 1.2
    return row
```

```python
df_transformed = df.apply(transform_row, axis="columns")
df_transformed[["Fare", "Fare_centered", "Fare_plus_20pct"]].head()
```
